```python
import socket

# Server configuration
HOST = "127.0.0.1"
PORT = 5055

# Create server socket
server = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

try:
    # Bind IP address and port
    server.bind((HOST, PORT))

    # Listen for client
    server.listen(1)

    print("SERVER STARTED SUCCESSFULLY", flush=True)
    print("Server IP:", HOST, flush=True)
    print("Port Number:", PORT, flush=True)
    print("Status: Waiting for client...", flush=True)

    # Accept client connection
    client_socket, client_address = server.accept()

    print("\nCLIENT CONNECTED!", flush=True)
    print("Client IP:", client_address[0], flush=True)
    print("Client Port:", client_address[1], flush=True)
    print("Status: Connected", flush=True)

    # Exchange 3 messages
    for i in range(3):

        # Receive message
        message = client_socket.recv(1024).decode("utf-8")

        if not message:
            print("Client disconnected.", flush=True)
            break

        print("\nClient:", message, flush=True)

        # Send response
        response = "Message received by server!"

        client_socket.sendall(response.encode("utf-8"))

        print("Server:", response, flush=True)

    print("\nChat completed!", flush=True)

    client_socket.close()

except OSError as e:
    print("Server Error:", e, flush=True)

finally:
    server.close()
    print("Server closed.", flush=True)
```
